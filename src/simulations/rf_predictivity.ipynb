# Tests How RFs perform on the Two Implemented Multiverses


```python
# imports
import pandas as pd
from sklearn.ensemble import RandomForestRegressor
from sklearn.metrics import r2_score, mean_squared_error
from sklearn.model_selection import train_test_split
```

```python
# read in the fertility and hurricane results
fertility = pd.read_csv("../../case_studies/fertility/output/runs/1/data/agg_1_run_outputs.csv.gz")
hurricane = pd.read_csv("../../case_studies/hurricane/output/runs/1/data/agg_1_run_outputs.csv.gz")
```

```python
# make the X columns the ones that start with mv_dim
fertility_X_cols = [col for col in fertility.columns if col.startswith("mv_dim")]
fertility_y_col = "fertility_low_estimate"
fertility_X = fertility[fertility_X_cols]
fertility_y = fertility[fertility_y_col]
# one-hot encode the columns of fertility_X, since they are all categorical
fertility_X = pd.get_dummies(fertility_X, columns=fertility_X_cols)
```

```python
# fit random forest regression model, look at training error
rf = RandomForestRegressor(n_estimators=100, random_state=42, bootstrap=False)
rf.fit(fertility_X, fertility_y)
fertility_y_pred = rf.predict(fertility_X)
print("R^2:", r2_score(fertility_y, fertility_y_pred))
print("MSE:", mean_squared_error(fertility_y, fertility_y_pred))
```

```python
hurricane_X_cols = [col for col in hurricane.columns if col.startswith("mv_dim")]
hurricane_y_col = "mean_deaths"
hurricane_X = hurricane[hurricane_X_cols]
hurricane_y = hurricane[hurricane_y_col]
# one-hot encode the columns of hurricane_X, since they are all categorical
hurricane_X = pd.get_dummies(hurricane_X, columns=hurricane_X_cols)
```

```python
rf = RandomForestRegressor(n_estimators=100, random_state=42, bootstrap=False)
rf.fit(hurricane_X, hurricane_y)
hurricane_y_pred = rf.predict(hurricane_X)
print("R^2:", r2_score(hurricane_y, hurricane_y_pred))
print("MSE:", mean_squared_error(hurricane_y, hurricane_y_pred))
```

```python
# now see how they perform on new data using train/test split
fertility_X_train, fertility_X_test, fertility_y_train, fertility_y_test = train_test_split(
    fertility_X, fertility_y, test_size=0.4, random_state=42
)
rf = RandomForestRegressor(n_estimators=100, random_state=42, bootstrap=False)
rf.fit(fertility_X_train, fertility_y_train)
fertility_y_test_pred = rf.predict(fertility_X_test)
print("R^2:", r2_score(fertility_y_test, fertility_y_test_pred))
print("MSE:", mean_squared_error(fertility_y_test, fertility_y_test_pred))

hurricane_X_train, hurricane_X_test, hurricane_y_train, hurricane_y_test = train_test_split(
    hurricane_X, hurricane_y, test_size=0.4, random_state=42
)
rf = RandomForestRegressor(n_estimators=100, random_state=42, bootstrap=False)
rf.fit(hurricane_X_train, hurricane_y_train)
hurricane_y_test_pred = rf.predict(hurricane_X_test)
print("R^2:", r2_score(hurricane_y_test, hurricane_y_test_pred))
print("MSE:", mean_squared_error(hurricane_y_test, hurricane_y_test_pred))
```
